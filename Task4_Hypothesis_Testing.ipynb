**import Library & Load Dataset**

```python
import pandas as pd
from scipy.stats import ttest_ind, t

df = pd.read_excel("Task1_Cleaned_Dataset.xlsx")

df.head()
```

```python
male_sales = df[df["Gender"] == "Male"]["Total_Sales"]
female_sales = df[df["Gender"] == "Female"]["Total_Sales"]

print("Male average sales:", round(male_sales.mean(), 2))
print("Female average sales:", round(female_sales.mean(), 2))
print("Male sample size:", len(male_sales))
print("Female sample size:", len(female_sales))
```

```python
t_stat, p_value = ttest_ind(
    male_sales,
    female_sales,
    equal_var=False
)

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))
```

```python
alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Average sales differ significantly by gender.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is no statistically significant difference in average sales by gender.")
```
